### Teste do Qui-Quadrado

- HO (Hipótese Nula): As duas variáveis são independentes.
- H1 (Hipótese Complementar): As duas variáveis não são indepententes.
- No teste, o objetivo é confirmar ou recusar a hipótese nula.
- Quando o valor de probabilidade H0 for menor que 0.05 (p-value) recusamos a hipótese nula e aceitamos a complementar.

```python
df_tabcount = pd.crosstab(df_churn["Churn"], df_churn["Contract"])
```

```python
from scipy.stats import chi2_contingency
```

```python
chi_scores = chi2_contingency(df_tabcount)
```

```python
chi_scores
```

```python
# Número decimal de uma notação científica
pd.set_option("display.float_format", lambda x: "%.15f" % x)
```

```python
scores_churn = pd.Series(chi_scores[0])
pvalues_churn = pd.Series(chi_scores[1])
```

```python
df_chi_churn = pd.DataFrame({
    "Qui2": scores_churn,
    "PValue": pvalues_churn
})
```

```python
df_chi_churn
```

As variáveis não são independentes. E pelo chi² alto, podemos afirmar que há uma alta correlação entre as variáveis.

### Técnica IQR 
- formula: 
 - Acha o 1º quartil e 3º quartil 
 - Depois faz 3º quartil - 1º quartil

```python
q1_total = df_churn_month["TotalCharges"].quantile(0.25)
q3_total = df_churn_month["TotalCharges"].quantile(0.75)
```

```python
iqr = q3_total - q1_total
iqr
```

```python
lim_inf = q1_total - (iqr * 1.5)
lim_sup = q3_total + (iqr * 1.5)
```

```python
lim_inf, lim_sup
```

```python
# Acha quantos dados com outliers tem 
len(df_churn_month[(df_churn_month["TotalCharges"] < lim_inf) | (df_churn_month["TotalCharges"] > lim_sup)])
```
