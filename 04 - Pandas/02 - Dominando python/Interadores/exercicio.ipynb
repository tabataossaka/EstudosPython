Enunciado 1: Escreva uma função que recebe um texto e extrai todas as URLs Presente. Considere que um URL pode começar com "http://" ou "https://"
e deve incluir um dominio  seguido por um sufixo (ex: ".com", ".org")

Enunciado 2: Crie uma funcao que valide senhas. Uma senha e considerada valida se tiver pelo menos 8 caracteres, pelo menos uma letra maiuscula, uma letra minuscula e um numero. Retorne True se a senha for valida e False caso contrario 

Enunciado 3: Implemente uma funcao que conte o numero de palavras em um texto. Considere palavras como sequencias de caracteres alfabeiticos (ou alfanumaricos) que sao separadas por espaco em branco ou pontuacao.

```python
import re
```

```python
def extrair_urls(texto):
    padrao_url = r'https?://[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}'
    urls = re.findall(padrao_url, texto)
    return urls
```

```python
texto_exemplo = """
Visite nosso site em https://www.exemplo.com para mais informações.
Você também pode acessar nosso blog em http://blog.exemplo.com.
Entre em contato conosco através do nosso portal seguro: https://secure.exemplo.com/login.
"""

urls_encontradas = extrair_urls(texto_exemplo)
print("URLs encontradas:", urls_encontradas)
```

```python
def validar_senhas(senhas):
    padrao_senha = r'^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$'
    return bool(re.match(padrao_senha, senhas))
```

```python
senhas = ["SenhaForte1!", "senhaFraca123", "SenhaForte2!", "SenhaFraca", "SenhaForte3@", "senhaFraca!"
          ]

for s in senhas:
    print(f"A Senha: {s} e Válida: {validar_senhas(s)}")
```

```python
def contar_palavras(texto):
    padrao_palavra = r'\b\w+\b'
    palavras = re.findall(padrao_palavra, texto)
    return len(palavras)
```

```python
texto_exemplo = "Ola, este é um exemplo de texto para contar palavras."
numero_de_palavras = contar_palavras(texto_exemplo)
print(f"Número de palavras no texto: {numero_de_palavras}")
```
