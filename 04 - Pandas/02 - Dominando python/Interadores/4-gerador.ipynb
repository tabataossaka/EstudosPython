Yield
- E uma palavra chave para criar geradores. Quando voce usa ele em uma funcao, voce transorma a duncao em um gerador, que e um tipo especial de um interador.
- Return: Quando e usado numa funcao, e encerrado e o valor e retornado, Chamadas subsequentes a funcao iniciam a execucao novamente do codigo.
- Yield: Pode ser chamada novamente e retormar a execucao do ponto onde foi paudada, permitindo a producao de uma sequendia de valores ao lonfo do tempo.

```python
def numeros_impares(n):
    for i in range(1, n+1, 2):
        yield i 
```

```python
for numero in numeros_impares(10):
    print(numero)
    
```
