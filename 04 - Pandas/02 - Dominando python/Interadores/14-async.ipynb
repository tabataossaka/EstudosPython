Async - Assincrona 

- Permite que você escreva código que pode ser executado de forma assíncrona, ou seja, sem bloquear a execução do programa enquanto aguarda por operações de I/O (entrada/saída) ou outras tarefas demoradas.
- Facilita a execução de múltiplas tarefas simultaneamente, melhorando a eficiência e a capacidade de resposta do programa.
- Importante para criacao de API 


```python
import asyncio
import aiohttp
```

```python
async def tarefa(nome, tempo):
    print(f"Tarefa {nome} iniciada")
    await asyncio.sleep(tempo)
    print(f"Tarefa {nome} concluída!")
```

```python
async def main():
    tarefas = [
        tarefa("A", 2),
        tarefa("B", 3),
        tarefa("C", 1)
    ]
    await asyncio.gather(*tarefas)
```

```python
asyncio.ensure_future(main())
```

```python
async def busca_usuario(session, usuario_id):
    url = f"https://api.github.com/users/{usuario_id}"
    async with session.get(url) as response:
        return await response.json()
```

```python
async def main():
    async with aiohttp.ClientSession() as session:
        usuario_id = [1,2,3,4,5]
        tarefas = [busca_usuario(session, uid) for uid in usuario_id]

        #Aguarda a conclusão de todas as tarefas
        usuarios = await asyncio.gather(*tarefas)
        for usuario in usuarios:
            print(usuario)
         
```

```python
await main() # posso usar await main() no jupyter notebook, mas no terminal não posso usar await, então uso asyncio.run(main())
```
