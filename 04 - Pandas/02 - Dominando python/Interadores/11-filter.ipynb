Filter
- E usada para filtrar elementos de um interavel com base em uma funcao que pode ser True ou False
- O objetivo e permitir apenas os elementos que satisfazem uma condicao.

```python
def e_par(x):
    return x % 2 == 0
```

```python
numeros = [2, 3, 4, 5, 6, 7, 8, 9, 10]
```

```python
numeros_pares = filter(e_par, numeros)
list(numeros_pares)
```

```python
produtos = [
    {"nome": "Produto A", "preco": 10.0, "estoque": True},
    {"nome": "Produto B", "preco": 20.0, "estoque": False},
    {"nome": "Produto C", "preco": 15.0, "estoque": True},
    {"nome": "Produto D", "preco": 30.0, "estoque": False},
]
```

```python
def produtos_em_estoque(produto):
    return produto["estoque"]
```

```python
produtos_disponiveis = filter(produtos_em_estoque, produtos)
list(produtos_disponiveis)
```

```python
transacoes = [200.0, -50.0, 100.0, -25.0, 75.0]
```

```python
def despesas(transacao):
    return transacao < 0
```

```python
despesas_filtradas = filter(despesas, transacoes)
list(despesas_filtradas)
```
