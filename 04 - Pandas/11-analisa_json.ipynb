```python
import pandas as pd
```

```python
df = pd.read_json("Dados/receitas.json", encoding="utf-8")
df
```

```python
# Analise Total de Ingredientes por receita
receitas_expandida = []
for receita in df['receitas']:
    total_ingredientes = len(receita['ingredientes'])
    receita_info = {
        'titulo': receita['titulo'],
        'tempo_preparo': receita['titulo'],
        'porcoes': receita['porcoes'],
        'total_ingredientes': total_ingredientes
        
    }
    receitas_expandida.append(receita_info)
```

```python
receitas_expandida
```

```python
df_expandido = pd.DataFrame(receitas_expandida)
df_expandido
```
