MAP
- Funcao que aplica uma funcao a todos os itens de um interalvel (uma lista, por exemplo)
- E uma maneira eficiente e consisa de transformar dados 

```python
def quadrado(x):
    return x ** 2
```

```python
numeros = [1, 2, 3, 4, 5]
```

```python
for n in numeros:
    print(f"O quadrado de {n} é {quadrado(n)}")
```

```python
# usando o map

resultados = map(quadrado, numeros)
list(resultados)
```

```python
def celsius_para_fahrenheit(c):
    return (c * 9/5) + 32
```

```python
temperaturas_celsius = [0, 20, 37, 100]
```

```python
temperaturas_fahrenheit = map(celsius_para_fahrenheit, temperaturas_celsius)
list(temperaturas_fahrenheit)
```

```python
# funcao para padrinizar os nomes de uma lista 
def normalizar_nome(nome):
    return nome.strip().lower().capitalize()
```

```python
nomes = ["  Alice  ", "BOB", "  Charlie  ", "MARIA"]
```

```python
nomes_normalizados = map(normalizar_nome, nomes)
list(nomes_normalizados)
```
